```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Step 1: Load Dataset
df = pd.read_csv('Unemployment in India.csv')
print("Shape:", df.shape)
print(df.head())

# Step 2: Clean Column Names
df.columns = df.columns.str.strip()
print("\nColumns:", df.columns.tolist())

# Step 3: Basic Info
print("\nBasic Statistics:")
print(df.describe())

# Step 4: Check Missing Values
print("\nMissing Values:")
print(df.isnull().sum())

# Step 5: Unemployment Rate Over Time
plt.figure(figsize=(12,5))
df_grouped = df.groupby('Date')[
    'Estimated Unemployment Rate (%)'].mean()
plt.plot(df_grouped.index,
         df_grouped.values,
         color='red', marker='o')
plt.title('Unemployment Rate Over Time in India')
plt.xlabel('Date')
plt.ylabel('Unemployment Rate (%)')
plt.xticks(rotation=45)
plt.tight_layout()
plt.savefig('unemployment_trend.png')
plt.show()

# Step 6: Covid-19 Impact
# Before Covid = before April 2020
# After Covid = after April 2020
df['Date'] = pd.to_datetime(df['Date'], dayfirst=True)
df['Period'] = df['Date'].apply(
    lambda x: 'After COVID-19'
    if x >= pd.Timestamp('2020-04-01')
    else 'Before COVID-19'
)

plt.figure(figsize=(8,5))
sns.boxplot(data=df,
            x='Period',
            y='Estimated Unemployment Rate (%)',
            palette=['green','red'])
plt.title('COVID-19 Impact on Unemployment Rate')
plt.savefig('covid_impact.png')
plt.show()

# Step 7: Region wise Analysis
plt.figure(figsize=(14,6))
region_avg = df.groupby('Region')[
    'Estimated Unemployment Rate (%)'].mean().sort_values()
region_avg.plot(kind='bar', color='orange')
plt.title('Average Unemployment Rate by Region')
plt.xlabel('Region')
plt.ylabel('Unemployment Rate (%)')
plt.xticks(rotation=45)
plt.tight_layout()
plt.savefig('region_analysis.png')
plt.show()

# Step 8: Seasonal Trends
df['Month'] = df['Date'].dt.month_name()
monthly_avg = df.groupby('Month')[
    'Estimated Unemployment Rate (%)'].mean()
plt.figure(figsize=(12,5))
monthly_avg.plot(kind='bar', color='purple')
plt.title('Monthly Seasonal Trends in Unemployment')
plt.xlabel('Month')
plt.ylabel('Unemployment Rate (%)')
plt.xticks(rotation=45)
plt.tight_layout()
plt.savefig('seasonal_trends.png')
plt.show()

print("\n Analysis Complete!")
print("Key Insight: COVID-19 significantly increased")
print("unemployment rates across all regions of India")
```
